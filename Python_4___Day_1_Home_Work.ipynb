```python
# 14. Identity Operators:
# Create two lists with same contents
list1 = [1, 2, 3]
list2 = [1, 2, 3]

# Check if they are the same object in memory
same_object = list1 is list2

print("Do list1 and list2 refer to the same object?", same_object)

# The is operator checks if two variables point to the exact same object
# in memory, not just if they have the same contents.

# So, even though list1 and list2 have the same elements, they are two separate
# list objects stored at different memory locations.

# TabErrorherefore, list1 is list2 → False.
```

```python
# 15. Identity Operators
# Assign values
x = 10
y = x

# Check if they refer to the same object
same_object = x is y

print("Do x and y refer to the same object?", same_object)


# As you know The is operator checks whether two variables point to the exact
# same object in memory.

# Here, y = x does not create a new object; it makes y reference the
# same integer object that x points to.

# Therefore, x is y → True.
```

Real-World Problem: Customer Discount Eligibility

Problem Statement:

A retail store has a loyalty program. Customers are eligible for a 15% discount on their total purchase if they meet one of the following conditions:

They are a member of the loyalty program (indicated by a boolean is_loyalty_member).
Their total purchase amount (purchase_amount) is over $200.
However, there is an additional condition: if a customer uses a special discount coupon (indicated by a boolean has_discount_coupon), they are not eligible for the loyalty program discount, even if they meet the other criteria.

Task:

Write a program that determines if a customer is eligible for the 15% loyalty program discount based on their loyalty membership status, purchase amount, and whether they have a discount coupon. Print a message indicating whether they are eligible and the final price if they are.

Steps to Solve:

Define Variables: Create variables to represent the customer's is_loyalty_member (boolean), purchase_amount (float), and has_discount_coupon (boolean). Assign sample values to these variables.
Check Discount Eligibility (Initial): Use a combination of comparison and logical operators (>=, or) to check if the customer meets the initial criteria for the loyalty discount (member OR purchase amount > $200).
Check for Coupon Conflict: Use a logical operator (and) to check if the customer is initially eligible and has a discount coupon.
Determine Final Eligibility: Use a logical operator (not) to negate the result of the coupon conflict check. If they were initially eligible AND did NOT have a coupon, they are finally eligible.
Apply Discount (if eligible): If the customer is finally eligible for the discount, calculate the discounted price.
Print Results: Use f-strings to print a clear message indicating whether the customer is eligible for the loyalty discount and, if so, their final discounted price.

```python
# Take input from the user
is_loyalty_member_input = input("Are you a loyalty program member? (yes/no): ").strip().lower()
purchase_amount = float(input("Enter your total purchase amount: $"))
has_discount_coupon_input = input("Do you have a discount coupon? (yes/no): ").strip().lower()

# Convert input to boolean
is_loyalty_member = True if is_loyalty_member_input == "yes" else False
has_discount_coupon = True if has_discount_coupon_input == "yes" else False

# Step 1: Check initial eligibility
initially_eligible = is_loyalty_member or purchase_amount > 200

# Step 2: Check coupon conflict
coupon_conflict = initially_eligible and has_discount_coupon

# Step 3: Determine final eligibility
finally_eligible = initially_eligible and not coupon_conflict

# Step 4: Apply discount if eligible
if finally_eligible:
    discounted_price = purchase_amount * 0.85  # 15% discount
    print(f"Congratulations! You are eligible for the loyalty discount.")
    print(f"Your final price after 15% discount is: ${discounted_price:.2f}")
else:
    print("Sorry, you are not eligible for the loyalty discount.")
    print(f"Your total price remains: ${purchase_amount:.2f}")

```
